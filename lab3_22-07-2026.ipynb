```python
import numpy as np

a = np.array([1, 2, 3, 4, 5])
b = np.array([[1, 2, 3], [4, 5, 6]])
print("1D Array:", a)
print("2D Array:\n", b)
print("\nZeros:\n", np.zeros((2, 3)))
print("Ones:\n", np.ones((2, 3)))
print("Identity Matrix:\n", np.eye(3))
print("Shape:", b.shape)
print("Size:", b.size)
print("Dimensions:", b.ndim)
print("Data type:", b.dtype)

c = np.arange(6)
print("Original:", c)
print("Reshaped:\n", c.reshape(2, 3))
print("Ravel:", c.ravel())
print("First element:", a[0])
print("Slice:", a[1:4])
print("2D element:", b[1, 2])

print("Sum:", np.sum(a))
print("Mean:", np.mean(a))
print("Median:", np.median(a))
print("Standard Deviation:", np.std(a))
print("Min:", np.min(a))
print("Max:", np.max(a))
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.arange(1, 6)
y = [10, 20, 15, 25, 30]
data = np.random.randn(100)

plt.figure(figsize=(10,8))

plt.subplot(3,3,1)
plt.plot(x, y, color='blue', linestyle='--', marker='o', label='Line')
plt.title("Line Plot")
plt.xlabel("X")
plt.ylabel("Y")
plt.grid()
plt.legend()

plt.subplot(3,3,2)
plt.bar(['A','B','C','D'], [5,7,3,8], color='green')
plt.title("Bar Chart")

plt.subplot(3,3,3)
plt.barh(['A','B','C'], [3,6,9], color='orange')
plt.title("Horizontal Bar")

plt.subplot(3,3,4)
plt.scatter(x, y, color='red')
plt.title("Scatter Plot")

plt.subplot(3,3,5)
plt.hist(data, bins=10, color='purple')
plt.title("Histogram")

plt.subplot(3,3,6)
sizes = [20,30,25,25]
labels = ['A','B','C','D']
plt.pie(sizes, labels=labels, autopct='%1.1f%%')
plt.title("Pie Chart")

plt.subplot(3,3,7)
plt.boxplot(data)
plt.title("Box Plot")

plt.subplot(3,3,8)
plt.fill_between(x, y, color='cyan', alpha=0.5)
plt.title("Area Plot")

plt.subplot(3,3,9)
plt.stem(x, y)
plt.title("Stem Plot")

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
data = {
    "ID": [1,2,3,4,5],
    "Name": ["A","B","C","D","E"],
    "Marks": [90, 85, None, 70, 95],
    "Grade": ["A","B","A","C","A"]
}
df = pd.DataFrame(data)

df.to_csv("students.csv", index=False)
print("CSV file created!\n")

df = pd.read_csv("students.csv")

print("Head:\n", df.head())
print("\nTail:\n", df.tail())
print("\nShape:", df.shape)
print("\nColumns:", df.columns)
print("\nInfo:\n")
print(df.info())
print("\nDescribe:\n", df.describe())

print("\nSingle Column:\n", df["Name"])
print("\nMultiple Columns:\n", df[["Name","Marks"]])

print("\nloc[0]:\n", df.loc[0])
print("\niloc[0]:\n", df.iloc[0])

print("\nMarks > 80:\n", df[df["Marks"] > 80])

df["Marks"].fillna(df["Marks"].mean(), inplace=True)
print("\nAfter Filling Missing Values:\n", df)

df["Bonus"] = df["Marks"] + 5
print("\nAfter Adding Bonus Column:\n", df)

df.drop("Bonus", axis=1, inplace=True)
print("\nAfter Dropping Bonus Column:\n", df)

df_sorted = df.sort_values("Marks")
print("\nSorted Data:\n", df_sorted)

print("\nGroupby Grade (Mean Marks):\n", df.groupby("Grade")["Marks"].mean())

df["Marks_plus10"] = df["Marks"].apply(lambda x: x+10)
print("\nAfter Apply Function:\n", df)

df.to_csv("students_updated.csv", index=False)
print("\nUpdated CSV saved!")
```

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

data = {
    "Hours": [1,2,3,4,5,6,7,8],
    "Marks": [10,20,30,40,50,60,70,80]
}
df = pd.DataFrame(data)
X = df[["Hours"]]    
y = df["Marks"]      

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

model = LinearRegression()
model.fit(X_train, y_train)

y_pred = model.predict(X_test)
print("MSE:", mean_squared_error(y_test, y_pred))
print("R2 Score:", r2_score(y_test, y_pred))

print("Prediction for 9 hours:", model.predict([[9]]))
```

```python
import numpy as np
from tensorflow import keras
from tensorflow.keras import layers

X = np.array([[1],[2],[3],[4],[5],[6]], dtype=float)
y = np.array([2,4,6,8,10,12], dtype=float)

model = keras.Sequential([
    layers.Dense(1, input_shape=[1])
])

model.compile(
    optimizer='adam',
    loss='mean_squared_error'
)

model.fit(X, y, epochs=200)
print("Prediction for 7:", model.predict([7]))
```

```python
import numpy as np
from tensorflow import keras
from tensorflow.keras import layers

# Data
X = np.array([[1],[2],[3],[4],[5],[6]], dtype=float)
y = np.array([2,4,6,8,10,12], dtype=float)

# Model
model = keras.Sequential([
    keras.Input(shape=(1,)),
    layers.Dense(1)
])

# Compile
model.compile(optimizer='adam', loss='mean_squared_error')

# Train
model.fit(X, y, epochs=200, verbose=0)

# Predict (FIXED)
print("Prediction for 7:", model.predict(np.array([[7]])))
```
