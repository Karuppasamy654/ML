```python
#find_s algorithm
def find_s(training_data):
    hypothesis=None
    for example in training_data:
        attributes=example[:-1]
        label=example[-1]
        if label=="Yes":
            if hypothesis is None:
                hypothesis=attributes.copy()
            else:
                for i in range(len(hypothesis)):
                  if hypothesis[i]!=attributes[i]:
                    hypothesis[i]="?"
    return hypothesis

def predict(hypothesis,test_example):
    for i in range(len(hypothesis)):
        if hypothesis[i]!="?" and hypothesis[i]!=test_example[i]:
            return "No"
        return "Yes"
        
training_data = [
    ["Sunny", "Warm", "Normal", "Strong", "Yes"],
    ["Sunny", "Warm", "High", "Strong", "Yes"],
    ["Rainy", "Cold", "High", "Strong", "No"],
    ["Sunny", "Warm", "High", "Weak", "Yes"]
]
final_hypothesis=find_s(training_data)
print("Final Hypothesis: ",final_hypothesis)

test1=['Sunny','Warm','High','Strong']
test2=['Rainy','Warm','High','Weak']
print("Prediction for test1: ",predict(final_hypothesis,test1))
print("Prediction for test2: ",predict(final_hypothesis,test2))
```

```python
#candidate elimination algorithm
import copy

def more_general(h1, h2):
    for x, y in zip(h1, h2):
        if x != "?" and (x != y):
            return False
    return True


def consistent(h, x):
    for i in range(len(h)):
        if h[i] != "?" and h[i] != x[i]:
            return False
    return True
def candidate_elimination(data):
    num_attributes=len(data[0])-1
    S = ["∅"] * num_attributes
    G = [["?"] * num_attributes]
    print("Initial S:",S)
    print("Initial G:",G)
    for example in data:
        x=example[:-1]
        label=example[-1]
        if label=="Yes":
            G = [g for g in G if consistent(g, x)]
            if S[0] == "∅":
                S = x.copy()
            else:
                for i in range(num_attributes):
                    if S[i] != x[i]:
                        S[i] = "?"
        else: 
            new_G = []
            for g in G:
                if consistent(g, x):
                    for i in range(num_attributes):
                        if g[i] == "?":
                            if S[i] != "∅" and S[i] != x[i]:
                                new_h = g.copy()
                                new_h[i] = S[i]
                                new_G.append(new_h)
                        elif g[i] != x[i]:
                            new_G.append(g)
                else:
                    new_G.append(g)
            G = []
            for h in new_G:
                if h not in G:
                    G.append(h)

    return S, G

def predict(x, S, G):
    def match_S(x):
        for i in range(len(S)):
            if S[i] != "?" and S[i] != x[i]:
                return False
        return True

    def match_G(x):
        for g in G:
            ok = True
            for i in range(len(g)):
                if g[i] != "?" and g[i] != x[i]:
                    ok = False
                    break
            if ok:
                return True
        return False

    s_match = match_S(x)
    g_match = match_G(x)

    if s_match and g_match:
        return "Yes (Accepted)"
    elif not g_match:
        return "No (Rejected)"
    else:
        return "Uncertain"
        
training_data = [
    ["Sunny", "Warm", "Normal", "Strong", "Warm", "Same", "Yes"],
    ["Sunny", "Warm", "High", "Strong", "Warm", "Same", "Yes"],
    ["Rainy", "Cold", "High", "Strong", "Warm", "Change", "No"],
    ["Sunny", "Warm", "High", "Strong", "Cool", "Change", "Yes"],
    ["Sunny", "Warm", "Normal", "Weak", "Warm", "Same", "Yes"],
    ["Rainy", "Warm", "Normal", "Weak", "Warm", "Same", "No"]
]


S, G = candidate_elimination(training_data)

print("\nFinal Specific Hypothesis S:", S)
print("Final General Hypotheses G:", G)

test_cases = [
    ["Sunny", "Warm", "Normal", "Strong", "Warm", "Same"],
    ["Rainy", "Cold", "High", "Strong", "Warm", "Change"],
    ["Sunny", "Cold", "High", "Strong", "Cool", "Change"],
    ["Sunny", "Warm", "Normal", "Weak", "Warm", "Same"]
]

print("\nPredictions:")
for t in test_cases:
    print(t, "->", predict(t, S, G))
```
