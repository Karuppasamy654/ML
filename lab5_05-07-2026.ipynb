```python
#candidate elimination algorithm
import copy

def more_general(h1, h2):
    for x, y in zip(h1, h2):
        if x != "?" and (x != y):
            return False
    return True


def consistent(h, x):
    for i in range(len(h)):
        if h[i] != "?" and h[i] != x[i]:
            return False
    return True
def candidate_elimination(data):
    num_attributes=len(data[0])-1
    S = ["∅"] * num_attributes
    G = [["?"] * num_attributes]
    print("Initial S:",S)
    print("Initial G:",G)
    for example in data:
        x=example[:-1]
        label=example[-1]
        if label=="Yes":
            G = [g for g in G if consistent(g, x)]
            if S[0] == "∅":
                S = x.copy()
            else:
                for i in range(num_attributes):
                    if S[i] != x[i]:
                        S[i] = "?"
        else: 
            new_G = []
            for g in G:
                if consistent(g, x):
                    for i in range(num_attributes):
                        if g[i] == "?":
                            if S[i] != "∅" and S[i] != x[i]:
                                new_h = g.copy()
                                new_h[i] = S[i]
                                new_G.append(new_h)
                        elif g[i] != x[i]:
                            new_G.append(g)
                else:
                    new_G.append(g)
            G = []
            for h in new_G:
                if h not in G:
                    G.append(h)

    return S, G

def predict(x, S, G):
    def match_S(x):
        for i in range(len(S)):
            if S[i] != "?" and S[i] != x[i]:
                return False
        return True

    def match_G(x):
        for g in G:
            ok = True
            for i in range(len(g)):
                if g[i] != "?" and g[i] != x[i]:
                    ok = False
                    break
            if ok:
                return True
        return False

    s_match = match_S(x)
    g_match = match_G(x)

    if s_match and g_match:
        return "Yes (Accepted)"
    elif not g_match:
        return "No (Rejected)"
    else:
        return "Uncertain"
        
training_data = [
    ["Sunny", "Warm", "Normal", "Strong", "Warm", "Same", "Yes"],
    ["Sunny", "Warm", "High", "Strong", "Warm", "Same", "Yes"],
    ["Rainy", "Cold", "High", "Strong", "Warm", "Change", "No"],
    ["Sunny", "Warm", "High", "Strong", "Cool", "Change", "Yes"],
    ["Sunny", "Warm", "Normal", "Weak", "Warm", "Same", "Yes"],
    ["Rainy", "Warm", "Normal", "Weak", "Warm", "Same", "No"]
]


S, G = candidate_elimination(training_data)

print("\nFinal Specific Hypothesis S:", S)
print("Final General Hypotheses G:", G)

test_cases = [
    ["Sunny", "Warm", "Normal", "Strong", "Warm", "Same"],
    ["Rainy", "Cold", "High", "Strong", "Warm", "Change"],
    ["Sunny", "Cold", "High", "Strong", "Cool", "Change"],
    ["Sunny", "Warm", "Normal", "Weak", "Warm", "Same"]
]

print("\nPredictions:")
for t in test_cases:
    print(t, "->", predict(t, S, G))
```

```python
import numpy as np
import matplotlib.pyplot as plt

X = np.array([1, 2, 3, 4, 5])
Y = np.array([2, 4, 5, 4, 5])
mean_x = np.mean(X)
mean_y = np.mean(Y)
m = np.sum((X - mean_x) * (Y - mean_y)) / np.sum((X - mean_x) ** 2)
b = mean_y - m * mean_x
print("Slope (m):", m)
print("Intercept (b):", b)

def predict(x):
    return m * x + b

Y_pred = predict(X)

plt.scatter(X, Y, label="Actual Data")
plt.plot(X, Y_pred, label="Regression Line")
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Linear Regression")
plt.legend()
plt.show()

x_new = 6
print("Prediction for x=6:", predict(x_new))
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

y_true = ["Yes", "No", "Yes", "No", "Yes"]
y_pred = ["Yes", "Yes", "Yes", "No", "No"]
TP = TN = FP = FN = 0

for actual, pred in zip(y_true, y_pred):
    if actual == "Yes" and pred == "Yes":
        TP += 1
    elif actual == "No" and pred == "No":
        TN += 1
    elif actual == "No" and pred == "Yes":
        FP += 1
    elif actual == "Yes" and pred == "No":
        FN += 1

print("Manual Calculation:")
print("TP:", TP, "FP:", FP)
print("FN:", FN, "TN:", TN)

cm = confusion_matrix(y_true, y_pred, labels=["Yes", "No"])
print("\nConfusion Matrix:\n", cm)


disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=["Yes", "No"])
disp.plot()

plt.title("Confusion Matrix")
plt.show()
```
