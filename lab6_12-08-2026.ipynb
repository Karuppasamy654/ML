```python
#Logistic Regression
import math

X=[1,2,3,4,5,6]
y=[0,0,1,1,1,1]
b0=0.0
b1=0.0
learning_rate=0.1
epochs=10000
n=len(X)

def sigmoid(z):
    return 1/(1+math.exp(-z))

for epoch in range(epochs):
    db0=0.0
    db1=0.0
    for i in range(n):
        z=b0+b1*X[i]
        p=sigmoid(z)
        error=p-y[i]
        db0+=error
        db1+=error*X[i]
    db0=db0/n
    db1=db1/n
    b0=b0-learning_rate*db0
    b1=b1-learning_rate*db1
    
mse=0.0
mae=0.0
for i in range(n):
    z=b0+b1*X[i]
    p=sigmoid(z)
    error=p-y[i]
    mse+=error**2
    mae+=abs(error)
mse=mse/n
mae=mae/n
print("MSE:",mse)
print("MAE:",mae)

def predict(x):
    z=b0+b1*x
    probability=sigmoid(z)
    if probability>=0.5:
        prediction=1
    else:
        prediction=0
    return prediction,probability

X_test=[7,2]
for x in X_test:
    prediction,probability=predict(x)
    print("Input:",x)
    print("Probability:",probability)
    print("Prediction:",prediction)

```

```python
#Logistic Regression
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error
X=[[1],[2],[3],[4],[5],[6]]
y=[0,0,1,1,1,1]
model=LogisticRegression()
model.fit(X, y)
X_test=[[7],[2]]
prediction=model.predict(X_test)
print("Prediction:",prediction)
mse = mean_squared_error([1,0], prediction)
mae = mean_absolute_error([1,0], prediction)
print("MSE:", mse)
print("MAE:", mae)
```

```python
#Lasso Regression
import math

X=[1,2,3,4,5,6]
y=[0,0,1,1,1,1]
b0=0.0
b1=0.0
learning_rate=0.1
lambda_l1=0.1
epochs=10000
n=len(X)
def sigmoid(z):
    return 1/(1+math.exp(-z))

def sign(x):
    if x>0:
        return 1
    elif x<0:
        return -1
    else:
        return 0

for epoch in range(epochs):
    db0=0.0
    db1=0.0
    for i in range(n):
        z=b0+b1*X[i]
        p=sigmoid(z)
        error=p-y[i]
        db0+=error
        db1+=error*X[i]
    db0=db0/n
    db1=db1/n
    db1=db1+lambda_l1*sign(b1)
    b0=b0-learning_rate*db0
    b1=b1-learning_rate*db1

print("Intercept b0:",b0)
print("Coefficient b1:",b1)
mse=0.0
mae=0.0

for i in range(n):
    z=b0+b1*X[i]
    p=sigmoid(z)
    error=p-y[i]
    mse+=error**2
    mae+=abs(error)

mse=mse/n
mae=mae/n
print("MSE:",mse)
print("MAE:",mae)

def predict(x):
    z=b0+b1*x
    probability=sigmoid(z)
    if probability>=0.5:
        prediction=1
    else:
        prediction=0
    return probability,prediction

X_test=[7,2]
for x in X_test:
    probability,prediction=predict(x)
    print("\nInput:",x)
    print("Probability:",probability)
    print("Prediction:",prediction)
```

```python
#Lasso Regression
from sklearn.linear_model import Lasso
from sklearn.metrics import mean_squared_error,mean_absolute_error
x=[[1],[2],[3],[4],[5]]
y=[2,4,6,8,10]
model=Lasso(alpha=0.01)
model.fit(x,y)
y_pred=model.predict(x)
mse=mean_squared_error(y,y_pred)
mae=mean_absolute_error(y,y_pred)
print("Actual: ",y)
print("Predicted: ",y_pred)
print("MSE: ",mse)
print("MAE: ",mae)
```

```python
#Ridge Regression
from sklearn.linear_model import Ridge
from sklearn.metrics import mean_squared_error,mean_absolute_error
x=[[1],[2],[3],[4],[5]]
y=[2,4,6,8,10]
model=Ridge(alpha=0.01)
model.fit(x,y)
y_pred=model.predict(x)
mse=mean_squared_error(y,y_pred)
mae=mean_absolute_error(y,y_pred)
print("Actual: ",y)
print("Predicted: ",y_pred)
print("MSE: ",mse)
print("MAE: ",mae)
```
